Audio to Text Generation

load dataset

```python
from pydub import AudioSegment

# Convert MP3 to WAV
sound = AudioSegment.from_mp3("introduceme.mp3")
sound.export("imtroduceme.wav", format="wav")

```

```python
import speech_recognition as sr

r = sr.Recognizer()
with sr.AudioFile("imtroduceme.wav") as source:  # Use the WAV file
    audio = r.record(source)
text = r.recognize_google(audio, language="eng")
print(text)
```

```python
# Sentiment Analysis
from textblob import TextBlob

blob = TextBlob(text)
sentiment = blob.sentiment.polarity
```

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.probability import FreqDist
nltk.download('punkt_tab')


tokens = word_tokenize(text)
freq_dist = FreqDist(tokens)

print(freq_dist.most_common(10))
```

```python
# Sentiment Analysis
from textblob import TextBlob

blob = TextBlob(text)
sentiment = blob.sentiment.polarity
sentiment
```

```python
sentiment_label = "positive" if sentiment > 0 else "negative" if sentiment < 0 else "neutral"
sentiment_label
```
